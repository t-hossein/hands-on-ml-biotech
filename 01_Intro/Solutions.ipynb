## 1.5 List comprehensions

A compact way to build a list from a loop, optionally with a condition:

```python
[expression for item in iterable if condition]
```

````python
squares = [x ** 2 for x in range(1, 6)]
print(squares)

even_squares = [x ** 2 for x in range(1, 11) if x % 2 == 0]
print(even_squares)

# Two loops: the left `for` is the outer loop
pairs = [(x, y) for x in range(1, 4) for y in range(1, 4) if x < y]
print(pairs)
````

## 1.6 Functions

`def` creates a function. Arguments can have **default values**. Variables created inside a function are **local**: they disappear when the function returns.

````python
def gc_content(sequence, as_percent=False):
    """Fraction of G and C bases in a DNA sequence."""
    sequence = sequence.upper()
    n_gc = sequence.count("G") + sequence.count("C")
    fraction = n_gc / len(sequence)
    if as_percent:
        return 100 * fraction
    return fraction

print(gc_content("ATGCGC"))
print(gc_content("ATGCGC", as_percent=True))
````

### ✏️ Exercise 2

1. Write a function `my_minimum(v)` that returns the smallest entry of a vector **without** using built-ins such as `min` or `np.minimum`. Use a loop. (And don't call any variable `min`!)
2. Test it on a vector of 10 random integers between 1 and 100. Note that `rng.integers(1, 101, size=10)` excludes the upper bound, so this gives 1 to 100.

````python
def my_minimum(v):
    smallest = float("inf")
    for element in v:
        if element < smallest:
            smallest = element
    return smallest

rng = np.random.default_rng(0)
test_vector = rng.integers(1, 101, size=10)
print(test_vector)
my_minimum(test_vector)
````

## 1.7 `*args` and `**kwargs`

## 1.10 Errors: reading tracebacks and `try/except`

When something breaks, Python prints a **traceback**. Read it **from the bottom up**: the last line says *what* went wrong, and the lines above it say *where*.

````python
# Remove the # from the last line to see a traceback, then read it bottom-up.
def average_expression(values):
    return sum(values) / len(values)

# average_expression([])   # ZeroDivisionError: division by zero
````

`try/except` lets you handle an error you **expect**. Catch the specific error type, never a bare `except:`. A bare `except` also hides the bugs you don't expect.

````python
raw_values = ["3.2", "4.8", "NA", "5.1"]   # "NA" is common in real data files

parsed = []
for text in raw_values:
    try:
        parsed.append(float(text))
    except ValueError:
        parsed.append(None)

print(parsed)
````

## 2.1 Why NumPy?

Python lists are flexible but slow for numbers. NumPy arrays store numbers in one block of memory and run operations in compiled code. That's called **vectorization**: operate on the whole array at once instead of looping.

````python
values = list(range(1_000_000))
values_array = np.arange(1_000_000)

print("Python loop:")
%timeit sum(v * 2 for v in values)
print("NumPy:")
%timeit (values_array * 2).sum()
````

## 2.4 Element-wise operations and broadcasting

Arithmetic on arrays works **element by element**. No loop needed.

````python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print(a + b)
print(a * b)
print(a ** 2)
print(np.log(b))
````

**Broadcasting**: when shapes differ, NumPy stretches the smaller array to match, if the shapes are compatible. Compatible means that, comparing shapes from the right, each pair of sizes is equal or one of them is 1.

Our running example is a small gene-expression table: 6 samples (rows) × 4 genes (columns), with a label per sample (0 = healthy, 1 = tumor). This `X` / `y` layout is exactly what every ML model expects.

````python
gene_names = np.array(["TP53", "BRCA1", "MYC", "GAPDH"])
X = np.array([
    [5.1, 2.0, 7.9, 10.2],
    [3.2, 1.1, 12.4, 10.0],
    [5.6, 2.3, 8.1, 9.8],
    [2.9, 0.8, 13.0, 10.4],
    [3.4, 1.0, 11.7, 10.1],
    [5.3, 2.2, 7.5, 9.9],
])
y = np.array([0, 1, 0, 1, 1, 0])
print("X:", X.shape, " y:", y.shape)

X_scaled = X * 2                     
gene_offsets = np.array([1, 0, -1, 0])
X_shifted = X + gene_offsets      
print(X_shifted[:2])
````

## 2.5 Indexing

### Slicing in 2D

`X[row, column]`. A `:` means "everything along this axis".

````python
print(X[0, 2]) # sample 0, gene 2
print(X[0]) # first sample (a row): all 4 genes
print(X[:, 0]) # first gene (a column): all 6 samples
print(X[1:3, :2]) # samples 1-2, genes 0-1
````

### Boolean masks

A comparison gives an array of `True`/`False`. Using it as an index keeps only the `True` positions.

````python
is_tumor = y == 1
print(is_tumor)
print(X[is_tumor]) # only tumor samples
print(X[:, 2][X[:, 2] > 10]) # MYC values above 10
````

### Fancy indexing (with integer arrays)

Index with a list or array of positions to pick exactly those elements, in that order.

````python
print(gene_names[[2, 0]]) # MYC, then TP53
print(X[[0, 5, 3]]) # samples 0, 5 and 3, in that order
````

**Shuffling** uses fancy indexing. Shuffle `X` and `y` with the **same** permutation so each sample keeps its label. A train/test split works like this underneath.

````python
rng = np.random.default_rng(42)
idx = rng.permutation(len(y))
print("idx:", idx)

X_shuffled = X[idx]
y_shuffled = y[idx]
print(y_shuffled)
````

## 2.6 `axis`

Reductions like `mean`, `sum` and `max` can work along one axis.

- `axis=0` goes **down the rows** and gives one value **per column** (per gene).
- `axis=1` goes **across the columns** and gives one value **per row** (per sample).

Rule of thumb: **the axis you name is the one that disappears.** `X` is `(6, 4)`; `X.mean(axis=0)` is `(4,)`.

````python
X.mean()
````

````python
X.mean(axis=0)
````

````python
X.mean(axis=1)
````

````python
X_centered = X - X.mean(axis=0)
````

````python
X_centered
````

### ✏️ Using a Boolean mask and `axis`, compute `tumor_means`: the mean expression **of each gene** across **tumor samples only**. Which gene changes most between healthy and tumor?

````python
tumor_means = X[y == 1].mean(axis=0)
healthy_means = X[y == 0].mean(axis=0)
difference = tumor_means - healthy_means
print("tumor:  ", tumor_means)
print("healthy:", healthy_means)
print("biggest change:", gene_names[np.argmax(np.abs(difference))])
````

## 2.7 Random numbers

Use a **generator** with a **seed**. The same seed gives the same "random" numbers every time, so your results are reproducible.

````python
rng = np.random.default_rng(seed=42)

print(rng.random(5)) # uniform on [0, 1)
print(rng.integers(1, 7, size=10)) # discrete uniform from 1 to 6 (dice)
print(rng.binomial(n=1, p=0.3, size=10)) # Bernoulli: 1 with probability 0.3
print(rng.choice(["hip", "hop"], size=5)) # pick from a list
print(rng.normal(loc=0, scale=1, size=3)) # normal distribution
````

You'll also see the older style `np.random.rand(...)`, `np.random.seed(...)` in tutorials and on the exercise sheet. It still works. For new code, prefer `default_rng`.

### ✏️ Exercise 3

A common trick to generate 100 samples of a binary variable with probability 0.7 for `True` (and 0.3 for `False`) is:

```python
np.random.rand(100) <= 0.7
```

**Explain why this trick produces the desired result.** Write your answer in the cell below, then run the experiment after it.

*Explanation:* `np.random.rand(100)` draws 100 independent values $U$ from the uniform distribution on $[0, 1)$. For a uniform variable, the probability of landing in an interval equals the interval's length, so $P(U \le 0.7) = 0.7$. The element-wise comparison `<= 0.7` turns each draw into `True` with probability 0.7 and `False` with probability 0.3. That is exactly a Bernoulli(0.7) sample. Over 100 draws we expect about 70 `True` values, but the exact count varies because the sample is random.
